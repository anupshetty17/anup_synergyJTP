```python
weights = np.zeros((X_train_np.shape[1], 1))
bias = 0.0

print("Weights shape:", weights.shape)
print("Initial bias:", bias)
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-np.clip(z, -500, 500)))
```

```python
sigmoid(np.array([-2, -1, 0, 1, 2]))
```

```python
def predict_probability(X, weights, bias):
    z = np.dot(X, weights) + bias
    return sigmoid(z)
```

```python
sample_predictions = predict_probability(
    X_train_np[:5],
    weights,
    bias
)

sample_predictions
```

```python
def compute_cost(y_true, y_pred):
    epsilon = 1e-15
    y_pred = np.clip(y_pred, epsilon, 1 - epsilon)

    cost = -np.mean(
        y_true * np.log(y_pred) +
        (1 - y_true) * np.log(1 - y_pred)
    )

    return cost
```

```python
initial_predictions = predict_probability(
    X_train_np,
    weights,
    bias
)

initial_cost = compute_cost(
    y_train_np,
    initial_predictions
)

initial_cost
```

```python
def train_model(X, y, learning_rate, epochs):
    m, n = X.shape

    weights = np.zeros((n, 1))
    bias = 0.0

    costs = []

    for epoch in range(epochs):
        predictions = predict_probability(X, weights, bias)

        dw = np.dot(X.T, predictions - y) / m
        db = np.sum(predictions - y) / m

        weights -= learning_rate * dw
        bias -= learning_rate * db

        if epoch % 100 == 0:
            predictions = predict_probability(X, weights, bias)
            cost = compute_cost(y, predictions)
            costs.append(cost)

    return weights, bias, costs
```

```python
numpy_weights, numpy_bias, training_costs = train_model(
    X_train_np,
    y_train_np,
    learning_rate=0.01,
    epochs=1000
)
```

```python
print("Initial cost:", initial_cost)
print("Final cost:", training_costs[-1])
```

```python
plt.figure(figsize=(8, 5))
plt.plot(range(0, 1000, 100), training_costs)
plt.xlabel('Epochs')
plt.ylabel('Cost')
plt.title('NumPy Logistic Regression Training')
plt.grid()
plt.show()
```
